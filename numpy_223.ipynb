```python
# Array Indexing & Slicing

arr = np.array([10, 20, 30, 40, 50])

print(arr[0])      # First element
print(arr[-1])     # Last element
print(arr[1:4])    # Slicing (from index 1 to 3)
print(arr[::])    # Every second element
```

```python
print(arr)
```

```python
print(arr[0::2]) # start:stop:step/jump
```

## **For 2D arrays:**

```python
arr2d = np.array([[1, 2, 3], [4, 5, 6]])


print(arr2d)
```

```python
print(arr2d[0, 1])  # Row 0, Column 1
print(arr2d[:, 1])  # All rows, Column 1
print(arr2d[1, :])  # Row 1, all columns
```

```python
print(arr2d[:,:-1])
```

```python
# Arithmetic Operations


a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print(a + b)  # Element-wise addition
print(a - b)  # Element-wise subtraction
print(a * b)  # Element-wise multiplication
print(a / b)  # Element-wise division
print(a ** 2) # Square of each element
```

```python
# Aggregations


arr = np.array([1, 2, 3, 4, 5])

print(np.sum(arr))  # Sum of elements
print(np.mean(arr)) # Mean (average)
print(np.max(arr))  # Maximum element
print(np.min(arr))  # Minimum element
print(np.std(arr))  # Standard deviation
```

```python
# Reshaping & Transposing


arr = np.array([1, 2, 3, 4, 5, 6])
print(arr.ndim)
reshaped = arr.reshape(2, 3)
print(reshaped)

print(reshaped.ndim)
```

```python
arr2d = np.array([[1, 2, 3], [4, 5, 6]])
print(arr2d.T)  # Transpose of matrix
```

# Stacking & Concatenation

```python
# Horizontal Stacking

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print(np.hstack((a, b)))
```

```python
# Vertical Stacking

a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])
print(np.vstack((a, b)))
```

# Mini Project example: Normalization

```python
data = np.array([50, 100, 40, 300,  150, 200])

normalized = (data - np.min(data)) / (np.max(data) - np.min(data))

print(normalized) # Scaling
```
