```python
import matplotlib.pyplot as plt
```

# Line Plot

```python
# Line Plot
X = [1, 2, 3, 4, 5]
Y = [10, 35, 55, 65, 85]

plt.plot(X, Y)
plt.show()
```

## Customizing the plot

```python
from matplotlib.legend import Legend
X = [1, 2, 3, 4, 5]
Y = [10, 35, 55, 65, 85]
plt.plot(X, Y, marker='o', linestyle='--', color='black', label='Marks')
plt.xlabel('X-axis- Hours')
plt.ylabel('Y-axis - Marks')
plt.title('Line Plot of Hours Vs Marks')
plt.grid()
plt.legend()
plt.show()
```

# Bar Chart

```python
# Bar Chart
X = ['Assam', 'Bihar', 'Chandigarh', 'Delhi']
Y = [40, 90, 55, 65]

plt.bar(X, Y)
plt.show()
```

```python
# Horizontal Bar Chart
X = ['Assam', 'Bihar', 'Chandigarh', 'Delhi']
Y = [40, 90, 55, 65]

plt.barh(X, Y)
plt.show()
```

```python
# Bar Chart
X = ['Assam', 'Bihar', 'Chandigarh', 'Delhi']
Y = [40, 90, 55, 65]

plt.bar(X, Y, color=['green','red','blue', 'brown'])
plt.xlabel('States')
plt.ylabel('Crime Rate')
plt.title('Crime Rate of Indian States')
plt.show()
```

# **Scatter Plot**

```python
x = [5, 7, 8, 7, 2, 17, 2, 9, 4, 11]
y = [99, 86, 87, 88, 100, 86, 103, 87, 94, 78]


plt.scatter(x, y)
plt.show()
```

```python
x = [5, 7, 8, 7, 2, 17, 2, 9, 4, 11]
y = [99, 86, 87, 88, 100, 86, 103, 87, 94, 78]

plt.scatter(x, y, color='red')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Scatter Plot')
plt.show()
```

```python
import pandas as pd
```

Learn more: https://github.com/lovnishverma/Python-Getting-Started/blob/main/Importing_Datasets_in_Google_Colab_using_pandas.ipynb

```python
df = pd.read_csv("https://raw.githubusercontent.com/lovnishverma/datasets/refs/heads/main/pandas.csv")

df
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
```

```python
df
```

```python
plt.plot(df['Age'], df['Score'])
plt.show()
```

```python
X = df['Age']
Y = df['Score']
plt.plot(X, Y, marker='o', linestyle='--', color='black', label='Score')
plt.xlabel('X-axis- Age')
plt.ylabel('Y-axis - Score')
plt.title('Line Plot of Age Vs Score')
plt.grid()
plt.legend()
plt.show()
```

# Saving A Plot as png or pdf

```python
plt.plot(df['Age'], df['Score'])
plt.savefig('lplot.png', dpi=600) # Always Save Before Show()
plt.show()
```

pie chart

```python
plt.pie(df['Score'], labels=df['Name'], autopct='%1.1f%%')
plt.title("Score Distribution")
plt.show()
```

**When to Use Matplotlib?**

• When creating simple and static visualizations.

• When you need detailed customization.

• Ideal for data exploration and reporting.

**Limitations**

• Not as aesthetically modern as Seaborn or Plotly.

• Requires more effort for advanced visualizations.
